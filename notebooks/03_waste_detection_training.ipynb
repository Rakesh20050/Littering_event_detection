```python
CLASS_NAMES = {
    0: "bottle",
    1: "can",
    2: "cup",
    3: "paper",
    4: "plastic",
    5: "carton",
    6: "glass",
    7: "other_waste",
}

print(CLASS_NAMES)
```

```python
def map_taco_category(name):
    name = name.lower().strip()

    # -------------------------
    # BOTTLE
    # -------------------------
    if (
        "plastic bottle" in name
        or name == "clear plastic bottle"
        or name == "glass bottle"
    ):
        return 0

    # -------------------------
    # CAN
    # -------------------------
    if (
        name == "drink can"
        or name == "food can"
        or name == "aerosol"
    ):
        return 1

    # -------------------------
    # CUP
    # -------------------------
    if (
        "plastic cup" in name
        or "paper cup" in name
        or "foam cup" in name
        or "glass cup" in name
    ):
        return 2

    # -------------------------
    # PAPER
    # -------------------------
    if (
        name == "normal paper"
        or "magazine paper" in name
        or "wrapping paper" in name
        or name == "tissues"
        or name == "paper bag"
        or name == "paper straw"
        or name == "toilet tube"
    ):
        return 3

    # -------------------------
    # CARTON
    # -------------------------
    if (
        "carton" in name
        or name == "pizza box"
    ):
        return 5

    # -------------------------
    # GLASS
    # -------------------------
    if (
        name == "broken glass"
        or name == "glass jar"
    ):
        return 6

    # -------------------------
    # PLASTIC
    # -------------------------
    plastic_keywords = [
        "plastic",
        "styrofoam",
        "crisp packet",
        "plastic film",
        "carrier bag",
        "food container",
        "plastic utensils",
        "garbage bag",
        "spread tub",
        "squeezable tube",
        "six pack rings",
        "polypropylene bag",
        "tupperware",
        "foam food container",
        "aluminium foil",
        "aluminium blister pack",
        "carded blister pack",
    ]

    if any(keyword in name for keyword in plastic_keywords):
        return 4

    # -------------------------
    # EVERYTHING ELSE
    # -------------------------
    return 7
```

```python
for category in taco["categories"]:
    original_name = category["name"]
    new_class = map_taco_category(original_name)

    print(
        f"{original_name:40s} -> "
        f"{new_class} ({CLASS_NAMES[new_class]})"
    )
```

```python
import shutil
import random

OUTPUT_DIR = PROJECT_ROOT / "data" / "processed" / "waste_yolo"

IMAGE_DIR = OUTPUT_DIR / "images"
LABEL_DIR = OUTPUT_DIR / "labels"

for split in ["train", "val", "test"]:
    (IMAGE_DIR / split).mkdir(parents=True, exist_ok=True)
    (LABEL_DIR / split).mkdir(parents=True, exist_ok=True)

print("Output:", OUTPUT_DIR)
```

```python
image_files = {}

for image_path in (TACO_DIR / "data").rglob("*"):
    if image_path.is_file():
        image_files[image_path.name] = image_path

print("Images found:", len(image_files))
```

```python
annotations_by_image = {}

for annotation in taco["annotations"]:
    image_id = annotation["image_id"]

    if image_id not in annotations_by_image:
        annotations_by_image[image_id] = []

    annotations_by_image[image_id].append(annotation)


images_by_id = {
    image["id"]: image
    for image in taco["images"]
}

category_id_to_yolo = {}

for category in taco["categories"]:
    category_id_to_yolo[category["id"]] = map_taco_category(
        category["name"]
    )

print("Images:", len(images_by_id))
print("Annotation groups:", len(annotations_by_image))
```

```python
# 70% train, 20% validation , 10% test
random.seed(42)

image_ids = list(images_by_id.keys())

random.shuffle(image_ids)

total = len(image_ids)

train_end = int(total * 0.70)
val_end = int(total * 0.90)

train_ids = image_ids[:train_end]
val_ids = image_ids[train_end:val_end]
test_ids = image_ids[val_end:]

print("Train:", len(train_ids))
print("Validation:", len(val_ids))
print("Test:", len(test_ids))
```

```python
def find_image(file_name):
    """
    Find the actual TACO image on disk.
    """
    
    candidate = TACO_DIR / "data" / file_name
    
    if candidate.exists():
        return candidate

    filename_only = Path(file_name).name

    if filename_only in image_files:
        return image_files[filename_only]

    return None


def convert_split(split_name, selected_ids):

    copied = 0
    skipped = 0
    annotations_written = 0

    for image_id in selected_ids:

        image_info = images_by_id[image_id]

        file_name = image_info["file_name"]

        width = image_info["width"]
        height = image_info["height"]

        source_image = find_image(file_name)

        if source_image is None:
            skipped += 1
            continue

        output_image = IMAGE_DIR / split_name / source_image.name

        shutil.copy2(
            source_image,
            output_image
        )

        label_path = (
            LABEL_DIR
            / split_name
            / f"{source_image.stem}.txt"
        )

        with open(label_path, "w", encoding="utf-8") as label_file:

            for annotation in annotations_by_image.get(image_id, []):

                category_id = annotation["category_id"]

                if category_id not in category_id_to_yolo:
                    continue

                x, y, box_width, box_height = annotation["bbox"]

                # Ignore invalid boxes
                if box_width <= 0 or box_height <= 0:
                    continue

                # COCO → YOLO
                x_center = (x + box_width / 2) / width
                y_center = (y + box_height / 2) / height

                normalized_width = box_width / width
                normalized_height = box_height / height

                yolo_class = category_id_to_yolo[category_id]

                label_file.write(
                    f"{yolo_class} "
                    f"{x_center:.6f} "
                    f"{y_center:.6f} "
                    f"{normalized_width:.6f} "
                    f"{normalized_height:.6f}\n"
                )

                annotations_written += 1

        copied += 1

    print(f"\n{split_name.upper()}")
    print("Images copied:", copied)
    print("Images skipped:", skipped)
    print("Annotations written:", annotations_written)
```

```python
convert_split("train", train_ids)
convert_split("val", val_ids)
convert_split("test", test_ids)
```

```python
from ultralytics import YOLO
import cv2

model = YOLO("yolo11n.pt")

camera_url = "http://130.1.8.118:8080/video"

cap = cv2.VideoCapture(camera_url)

if not cap.isOpened():
    print("Could not connect to phone camera.")
else:
    print("Phone camera connected.")
    print("Starting ByteTrack...")

    for frame_number in range(30):

        success, frame = cap.read()

        if not success:
            continue

        results = model.track(
            source=frame,
            persist=True,
            tracker="bytetrack.yaml",
            conf=0.25,
            verbose=False
        )

        boxes = results[0].boxes

        if boxes is None or len(boxes) == 0:
            print(f"Frame {frame_number}: no detections")
            continue

        for box in boxes:

            class_id = int(box.cls[0])
            class_name = model.names[class_id]
            confidence = float(box.conf[0])

            if class_name != "person":
                continue

            if box.id is not None:
                track_id = int(box.id[0])

                print(
                    f"Frame {frame_number}: "
                    f"PERSON | "
                    f"Track ID={track_id} | "
                    f"confidence={confidence:.3f}"
                )
            else:
                print(
                    f"Frame {frame_number}: "
                    f"PERSON | "
                    f"No Track ID"
                )

    cap.release()

    print()
    print("ByteTrack test completed.")
```

```python
from ultralytics import YOLO
import cv2

model = YOLO("yolo11n.pt")

camera_url = "http://130.1.8.118:8080/video"

cap = cv2.VideoCapture(camera_url)

if not cap.isOpened():
    print("Could not connect to phone camera.")
else:
    print("Phone camera connected.")
    print("ByteTrack running. Press Q to stop.")

    while True:

        success, frame = cap.read()

        if not success:
            print("Frame read failed.")
            break

        results = model.track(
            source=frame,
            persist=True,
            tracker="bytetrack.yaml",
            conf=0.10,
            verbose=False
        )

        annotated = results[0].plot()

        boxes = results[0].boxes

        if boxes is not None and len(boxes) > 0:

            for box in boxes:

                class_id = int(box.cls[0])
                class_name = model.names[class_id]
                confidence = float(box.conf[0])

                if class_name == "person":

                    if box.id is not None:
                        track_id = int(box.id[0])

                        print(
                            f"PERSON | "
                            f"Track ID={track_id} | "
                            f"confidence={confidence:.3f}"
                        )

        cv2.imshow(
            "YOLO11 + ByteTrack",
            annotated
        )

        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    cap.release()
    cv2.destroyAllWindows()

    print("ByteTrack test completed.")
```
